```python
from pyannote.audio import Pipeline
import torch

from dotenv import load_dotenv

load_dotenv()

# 替換為你的 Hugging Face 權杖
# 注意：v3.1 是最新的穩定版本，如果遇到問題可以嘗試 v2.1
pipeline = Pipeline.from_pretrained("pyannote/speaker-diarization-3.1",
                                    use_auth_token=HUGGINGFACE_API_KEY)

# 將 pipeline 移至 GPU (如果可用)
pipeline.to(torch.device("cuda"))

# 指定你的 WAV 檔案路徑
audio_file_path = "S01_P01.wav"
# 如果你的 WAV 檔案在 Colab session 中上傳，路徑可能是 "/content/my_audio.wav"
# audio_file_path = "/content/my_audio.wav"

# 執行講者辨識
diarization = pipeline(audio_file_path)

# 打印辨識結果
print("Speaker Diarization Results:")
for turn, _, speaker in diarization.itertracks(yield_label=True):
    print(f"start={turn.start:.1f}s stop={turn.end:.1f}s speaker={speaker}")
```

```python
import whisper

# 載入 Whisper 模型
# "large-v3" 是目前最大且最準確的模型，但需要更多計算資源
# "medium", "small", "base", "tiny" 則較快但準確度可能較低
# 如果記憶體不足，可以嘗試較小的模型
model = whisper.load_model("large-v3")

# 執行語音轉文字
# 這個步驟會將整個音檔轉成文字，沒有講者標記
result = model.transcribe(audio_file_path, language="en") 

print("\nFull Transcript:")
print(result["text"])
```

```python
# 這個部分會比較複雜，需要結合 Diarization 和 Whisper 的結果
# 簡單的實現方式：
# 遍歷 diarization 的結果，對每個講者說話的片段進行 Whisper 轉錄

from pydub import AudioSegment
from pydub.silence import split_on_silence # 用於切割音訊片段，可能會有幫助

# 載入原始音訊
audio = AudioSegment.from_wav(audio_file_path)

print("\nDetailed Transcript with Speakers:")
transcript_with_speakers = []

for turn, _, speaker in diarization.itertracks(yield_label=True):
    start_ms = turn.start * 1000  # 轉換為毫秒
    end_ms = turn.end * 1000      # 轉換為毫秒

    # 提取講者說話的音訊片段
    speaker_segment = audio[start_ms:end_ms]

    # 將片段暫存為 WAV 檔案以便 Whisper 處理
    temp_speaker_audio_path = f"/content/{speaker}_{int(start_ms)}.wav"
    speaker_segment.export(temp_speaker_audio_path, format="wav")

    # 使用 Whisper 轉錄該片段
    segment_result = model.transcribe(temp_speaker_audio_path, language="en")
    segment_text = segment_result["text"].strip()

    if segment_text: # 只有在有文字內容時才添加
        transcript_with_speakers.append(f"[{turn.start:.1f}s - {turn.end:.1f}s] {speaker}: {segment_text}")

# 打印最終的逐字稿
for line in transcript_with_speakers:
    print(line)

# 你也可以將結果寫入檔案
output_filename = "/content/drive/MyDrive/transcript_with_speakers.txt"
with open(output_filename, "w", encoding="utf-8") as f:
    for line in transcript_with_speakers:
        f.write(line + "\n")
print(f"\nTranscript saved to: {output_filename}")
```
